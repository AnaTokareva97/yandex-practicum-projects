## Исследовательский анализ данных

### Запрос №1

Посчитаем, сколько книг было выпущено после 1 января 2000 года.

```python
#считаем книги, выпущенные после 01.01.2000
query = '''
SELECT COUNT(book_id)
FROM books
WHERE publication_date > '2000-01-01'
'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

После 1 января 2000 года было выпущено 819 книг.

### Запрос №2

Для каждой книги посчитаем количество обзоров и среднюю оценку.

```python
#считаем количество обзоров и среднюю оценку для каждой книги
query = '''
SELECT b.title,
       COUNT(DISTINCT rev.review_id) AS review_cnt,
       AVG(rat.rating) AS average_rating
FROM books AS b
LEFT JOIN reviews AS rev ON b.book_id = rev.book_id
LEFT JOIN ratings AS rat ON b.book_id = rat.book_id
GROUP BY b.title
'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

Мы вывели все книги и к каждой количество обзоров и средний рейтинг.

Больше всего обзоров у книги с названием Memoirs of a Geisha - 8 обзоров.

Книг с самым высоким рейтингом 5.0 44 штуки. Но среди таких книг есть одна, у которой также больше всего обзоров (4), -  A Dirty Job (Grim Reaper #1).

### Запрос №3

Определим издательство, которое выпустило наибольшее число книг толще 50 страниц. Таким образом мы исключим из анализа брошюры.

```python
#найдем издательство, которое выпустило наибольшее число книг толще 50 страниц
query = '''
SELECT DISTINCT p.publisher,
       COUNT(b.book_id) AS books_cnt
FROM books AS b
JOIN publishers AS p ON b.publisher_id = p.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher
ORDER BY books_cnt DESC
LIMIT 1
'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

Издательство Penguin Books выпустили самое большое количество книг толще 50 страниц - 42 книги.

### Запрос №4

Определим автора с самой высокой средней оценкой книг, у которых 50 и более оценок.

```python
#найдем такого автора
query = '''
WITH author_rating AS (SELECT b.book_id,
                              b.author_id,
                              AVG(rat.rating) AS average_rating,
                              COUNT(rat.rating_id) AS rating_count
                       FROM books AS b
                       JOIN ratings AS rat ON b.book_id = rat.book_id
                       GROUP BY 1, 2
                       HAVING COUNT(rat.rating_id) >= 50)
SELECT a.author,
       AVG(ar.average_rating) AS avg_rating
FROM authors AS a
JOIN author_rating AS ar ON a.author_id = ar.author_id
GROUP BY 1
ORDER BY 2 DESC
LIMIT 1
'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

Автором с самым высоким средним рейтингов, среди книг с 50 и более оценками, оказалась J.K. Rowling и др. Средний рейтинг - 4,28.

### Запрос №5

Посчитаем среднее количество обзоров от пользователей, которые поставили больше 48 оценок.

```python
query = '''
WITH big_rating AS (SELECT username,
                           COUNT(rating_id) AS rating_cnt
                    FROM ratings
                    GROUP BY 1
                    HAVING COUNT(rating_id) > 48),
     reviews_quantity AS (SELECT username,
                                 COUNT(review_id) AS review_cnt
                          FROM reviews
                          GROUP BY 1)
SELECT AVG(rq.review_cnt) AS avg_reviews
FROM big_rating AS br
LEFT JOIN reviews_quantity AS rq ON br.username = rq.username
'''
con=engine.connect()
pd.io.sql.read_sql(sql=text(query), con = con)
```

В среднем пользователи, которые поставили больше 48 оценок, оставляют 24 обзора на книги.
